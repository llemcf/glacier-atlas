### Functions

```python
def models_avg_dataset(model_list, scenario):
    dicts_nmb_xtr_days = {}
    dicts_xtr_days = {}
    
    s3f = s3fs.S3FileSystem()
    
    for m in model_list:
        if scenario == "ssp585":
            zarr_path = f"s3://deepki/opendata/NEX-GDDP-CMIP6/precipitation/{m}/zarr/ds_kpi_ssp85/"
        else : 
            zarr_path = f"s3://deepki/opendata/NEX-GDDP-CMIP6/precipitation/{m}/zarr/ds_kpi/"
        store = s3fs.S3Map(root=zarr_path, s3=s3f, check=False)
        ds = xr.open_zarr(store=store, consolidated=True)
        dicts_nmb_xtr_days[m] = ds.nmb_xtr_days.squeeze()
        dicts_xtr_days[m] = ds.rain_xtr_days.squeeze()

    nmb_xtr_days = xr.Dataset(dicts_nmb_xtr_days)
    rain_xtr_days = xr.Dataset(dicts_xtr_days)
    
    mean_nmb_xtr_days = nmb_xtr_days.to_array(dim='models').mean('models')
    mean_rain_xtr_days = rain_xtr_days.to_array(dim='models').mean('models')
    
    nmb_xtr_days = nmb_xtr_days.assign(avg=mean_nmb_xtr_days)
    rain_xtr_days = rain_xtr_days.assign(avg=mean_rain_xtr_days)
    
    ds_indicators = xr.Dataset(
        data_vars=dict(
        nmb_xtr_days=(["lat", "lon"], nmb_xtr_days.avg.data),
        rain_xtr_days=(["lat", "lon"], rain_xtr_days.avg.data),
        ),
        coords=dict(
        lon = nmb_xtr_days.lon,
        lat = nmb_xtr_days.lat
        ))
   
    #ds_indicators.to_netcdf(f"../docs/ds_indicators_best_model_avg_{scenario}.nc")

    return ds_indicators
```

```python
def get_indicators_for_the_total_historical_period(ds: xr.Dataset) -> xr.Dataset:
    ds["avg_hw_day_temp_hist"] = (ds["avg_hw_day_temp_81_91"] + ds["avg_hw_day_temp_91_01"] + ds["avg_hw_day_temp_01_11"]) / 3
    ds["nb_hw_days_hist"] = (ds["nb_hw_days_81_91"] + ds["nb_hw_days_91_01"] + ds["nb_hw_days_01_11"]) / 3
    return ds

def adding_historical_kpis(ds: xr.Dataset, s3_path_hist: str) -> xr.Dataset:
    ds_hist = get_model(s3_path_hist)
    ds_hist = get_indicators_for_the_total_historical_period(ds_hist)
    ds_merged = xr.merge([ds, ds_hist])
    
    return ds_merged

def compute_mean_dataset(ds: xr.Dataset) -> xr.Dataset:
    """ Computes the mean of all of a dataset variables and returns it in the same dataset with a new 'avg' variable """
    
    mean = ds.to_array(dim='new').mean('new')
    ds = ds.assign(avg=mean)
    
    return ds
    
def get_mean_dataset(list_models: List, scenario : Literal["ssp245", "ssp585"] = "ssp245"):
    ''' Based on a list of models, returns a dataset with the 'avg_hw_day_temp' and 'nb_hw_days' variables averaged for the whole historical period (1981-2010).
    Those datasets are generated thanks to the script rnd-explo/resilience_worldwide/load_models_histo_kpi.py
    '''
    
    dic_da = {}
    for i, model in enumerate(list_models):
        s3_path = "opendata/NEX-GDDP-CMIP6/heatwaves/{}/zarr/ds_summary_{}_vOuzeau_adapted_clean_10"
        s3_path_hist = "s3://deepki/" + s3_path.format(model, 'historical')
        s3_path = "s3://deepki/" + s3_path.format(model, scenario)
        ds = get_model(s3_path)
        ds = adding_historical_kpis(ds, s3_path_hist)

        for var in ds.data_vars:
            if i == 0:
                dic_da[var] = {}
            if "temp" in var:
                ds[f"{var}"] = ds[f"{var}"].where(ds[f"{var}"] != 0, np.nan) # for models that have identified no hw at a given location we ignore their hw days temp that is set to 0 (and hence distort the mean)
            dic_da[var][model] = ds[f"{var}"]

    dic_final_ds = {}

    for var in dic_da:
        ds_var = xr.Dataset(dic_da[var])
        ds_var = compute_mean_dataset(ds_var)
        dic_final_ds[var] = ds_var.avg

    ds_mean = xr.Dataset(dic_final_ds)
    
    return ds_mean
```

```python
def get_model(zarr_path):
    s3f = s3fs.S3FileSystem()
    store = s3fs.S3Map(root=zarr_path, s3=s3f, check=False)
    ds = xr.open_zarr(store=store, consolidated=True)
    
    return ds
```

```python
def compute_cont_score(continent, risk, scenario, ds_point):
    scale_axis = params_dict[continent][risk]["scale_axis"][scenario]
    ds_point[f"cont_score"] = np.sqrt(ds_point["ind_1"]**2 + (ds_point["ind_2"] * scale_axis)**2)
    return ds_point
```

```python
def attribute_score(value: float, breakpoints: List[float], scores="012345") -> int:
    """
    Function to discretize a continuous value
    """
    if not math.isnan(value):
        i = bisect(breakpoints, value)
        return int(scores[i])+1
    return int(1)#int(-1)
```

```python
def attribute_score_dd(value: float, breakpoints: List[float], scores="012345") -> int:
    """
    Function to discretize a continuous value
    """
    if not math.isnan(value):
        i = bisect(breakpoints, value)
        return int(scores[i])
    return int(-1)
```

```python
def fetch_scores_precipitations(building, scenario): 
    
    if continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))].values.shape[0] == 0:
        discrete_score = "lat/lon out of range"
    else:
        continent = continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))]["continent"].values[0]
        ds_precipitations = models_avg_dataset(model_lists[risk], scenario)
            
        build_sel = ds_precipitations.sel(lat = building["lat"], lon = building["lon"], method="nearest").compute().rename({"nmb_xtr_days" : "ind_1", "rain_xtr_days" : "ind_2"})
        build_sel = compute_cont_score(continent, risk, scenario, build_sel)
    
        discrete_score = attribute_score(build_sel.cont_score, params_dict[continent][risk]["bounds"])
    return discrete_score
```

```python
def fetch_scores_clay_subsidence(building, scenario): 
    if continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))].values.shape[0] == 0:
        discrete_score = "lat/lon out of range"
    else:    
        continent = continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))]["continent"].values[0]
    
        if scenario == "ssp245":
            ds_clay_subsidence = get_model(f"s3://deepki/opendata/rga/ds_summary_45_{continent}/")
    
        elif scenario == "ssp585":
            ds_clay_subsidence = get_model(f"s3://deepki/opendata/rga/ds_summary_85_{continent}/")
    
        build_sel = ds_clay_subsidence.sel(y = building["lat"], x = building["lon"], method="nearest").compute().rename({"clay" : "ind_1", "max_min_delta" : "ind_2", "x" : "lon", "y" : "lat"})
        build_sel = ds_clay_subsidence.sel(y = building["lat"], x = building["lon"], method="nearest").compute().rename({"clay" : "ind_1", "max_min_delta" : "ind_2", "x" : "lon", "y" : "lat"})
        
        build_sel = compute_cont_score(continent, risk, scenario, build_sel)
    
        discrete_score = attribute_score(build_sel.cont_score, params_dict[continent][risk]["bounds"])

    return discrete_score
```

```python
def fetch_scores_dry_days(building, scenario): 
    if continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))].values.shape[0] == 0:
        discrete_score = "lat/lon out of range"
    else:    
        continent = continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))]["continent"].values[0]
        ds_dry_days = get_model(f"s3://deepki/opendata/NEX-GDDP-CMIP6/precipitation/dry_days_temp/ds_indicators_dry_days_{scenario}/")

        if scenario == "ssp245":
            build_sel = ds_dry_days.sel(lat = building["lat"], lon = building["lon"], method="nearest").compute().rename({"max_cdd_evl" : "ind_1", "max_cdd_2050" : "ind_2"}) 
        elif scenario == "ssp585":
            build_sel = ds_dry_days.sel(lat = building["lat"], lon = building["lon"], method="nearest").compute().rename({"evl_hist_2050" : "ind_1", "max_cdd_2050" : "ind_2"}) 
       
        build_sel = compute_cont_score(continent, risk, scenario, build_sel)
        
        discrete_score = attribute_score_dd(build_sel.cont_score, params_dict[continent][risk]["bounds"])
    return discrete_score
```

```python
def fetch_scores_heatwaves(building, scenario): 
    if continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))].values.shape[0] == 0:
        discrete_score = "lat/lon out of range"
    else:    
        continent = continent_shapes.loc[continent_shapes.geometry.contains(Point(building["lon"], building["lat"]))]["continent"].values[0]
    
        ds_heatwaves = get_mean_dataset(model_lists[risk], scenario)
    
        build_sel = ds_heatwaves.sel(lat = building["lat"], lon = building["lon"], method="nearest").compute().rename({"avg_hw_day_temp_45_55" : "ind_1", "nb_hw_days_45_55" : "ind_2"})
        
        build_sel = compute_cont_score(continent, risk, scenario, build_sel)
        build_sel["cont_score"] = build_sel.cont_score.where(build_sel.ind_2 != 0, 0)
    
        discrete_score = attribute_score_dd(build_sel.cont_score, params_dict[continent][risk]["bounds"])

    return discrete_score
```
