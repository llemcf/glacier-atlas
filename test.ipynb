```python
import os
fichiers = []
dossier = '/Users/guillem.carcanade/Documents/guillem/glacier_atlas/data/images'
for root, dirs, files in os.walk(dossier):
    for name in files:
        fichiers.append(name)
```

```python
import geopandas as gpd
path = "data/randolph_glacier_inventory_7/rgi2000_v70_vector.shp"
```

```python
import re
import csv
fichiers.sort()

rows = []

for nom_fichier in fichiers:
    # Recherche d'une année commençant par 19 ou 20
    match_date = re.search(r'(19|20)\d{2}', nom_fichier)
    
    annee = ""
    etat = ""
    
    if match_date:
        annee = match_date.group(0)
        # Logique d'état
        if annee.startswith("19"):
            etat = "old"
        elif annee.startswith("20"):
            etat = "new"
    else:
        annee = "Non trouvée"
        etat = "Inconnu"
        
    rows.append([nom_fichier, annee, etat])

# Écriture du fichier CSV
output_filename = 'analyse_images.csv'
with open(output_filename, mode='w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Nom du fichier', 'Date', 'Etat']) # En-têtes
    writer.writerows(rows)

print(f"Fichier '{output_filename}' généré avec succès ({len(rows)} lignes).")
```

```python
from pathlib import Path

def afficher_arborescence(dossier: Path, prefix=""):
    # On récupère tous les éléments du dossier, triés par nom
    elements = sorted(dossier.iterdir(), key=lambda x: x.name)
    
    # On compte pour savoir quand on est au dernier élément (pour changer le symbole)
    total = len(elements)
    
    for index, chemin in enumerate(elements):
        is_last = (index == total - 1)
        
        # Choix du connecteur graphique
        connector = "└── " if is_last else "├── "
        
        # Affichage
        print(f"{prefix}{connector}{chemin.name}")
        
        # Si c'est un dossier, on plonge dedans (récursivité)
        if chemin.is_dir():
            # On prépare le préfixe pour les enfants
            extension = "    " if is_last else "│   "
            afficher_arborescence(chemin, prefix + extension)

# --- Utilisation ---
dossier_racine = Path('.') # Dossier courant
print(f"📁 {dossier_racine.resolve().name}")
afficher_arborescence(dossier_racine)
```
